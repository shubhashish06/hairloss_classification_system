```python
import pandas as pd
import shutil
import os
```

```python
#loading male annotation file

data_set=pd.read_csv('/Users/ashish/Desktop/final_projects/hairloss_classification/Anno/male_annotations.csv')
```

```python
#In the given dataset there's a special column to mention whether a person is bald or not [1,-1] showing that 5.3% of males are bald
#It can be concluded that person with annot Bald=1 are already at their stage 4, so lets classify them into stage-4 and put them into an external folder
print("Total no. of males:",len(data_set[data_set['Bald']==1]))
print("No. of males who are bald:",len(data_set))
k=len(data_set[data_set['Bald']==1])
l=len(data_set)
print(k/l)
```

```python
#In the given data-set it can be easily concluded that a person has a 11% chance of having a receding hairline
m=len(data_set[data_set['Receding_Hairline']==1])
print("No. of males who have receding hairline:",m)
print(m/l)
```

```python

#putting them into seperate folders named stage(0,1,2)
source='/Users/ashish/Desktop/final_projects/hairloss_classification/male_images'
dest1='/Users/ashish/Desktop/final_projects/hairloss_classification/stage0'
dest2='/Users/ashish/Desktop/final_projects/hairloss_classification/stage1'
dest3='/Users/ashish/Desktop/final_projects/hairloss_classification/stage2'
```

```python
count1 = count2 = count3 = 0
for index, row in data_set.iterrows():
    if row['Receding_Hairline'] == -1 and row['Bald'] == -1 and count1 < 500:
        des = dest1
        count1 += 1
        shutil.copy2(os.path.join(source, row['file_name']), des)
    
    elif row['Receding_Hairline'] == 1 and row['Bald'] == -1 and count2 < 500:
        des = dest2
        count2 += 1
        shutil.copy2(os.path.join(source, row['file_name']), des)

    elif row['Receding_Hairline'] == 1 and row['Bald'] == 1 and count3 < 500:
        des = dest3
        count3 += 1
        shutil.copy2(os.path.join(source, row['file_name']), des)
```

```python
#we manually removed the images from each class that are either misclassified or were not good enough to capture patterns such as
# 1. wearing caps 2. blurry images 3. black background 3. mis-classified female images
```
